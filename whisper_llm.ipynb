```python
import mlx_whisper

# Define your audio file path
audio_file = "/Users/marcignacio/Downloads/2026JAN23.mp3"

# Transcribe using the "large-v3-turbo" model 
# (This is the best balance of extremely high accuracy and speed for Taglish)
result = mlx_whisper.transcribe(
    audio_file, 
    path_or_hf_repo="mlx-community/whisper-large-v3-turbo"
)

# Print the text
print(result["text"])

# Save to a text file
with open("transcription.txt", "w", encoding="utf-8") as f:
    f.write(result["text"])
```
